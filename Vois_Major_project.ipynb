# Data Cleaning & Imputation Strategy
**Impute numerical missing values with median and Verify missing values after cleaning**



```python
for col in ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())

print("Missing values remaining:", df.isna().sum().sum())
```

**Cost-to-Profit Quadrant Analysis & Return on Investment (ROI) Distribution**

```python
# Calculate ROI percentage if it doesn't already exist in the dataframe
df['ROI_pct'] = (df['Profit_INR'] / df['Total_Cost_INR']) * 100

# Set up the figure dimensions
plt.figure(figsize=(10, 6))

# Create a scatter plot mapping Total Cost vs. Profit, colored and sized by ROI
scatter = sns.scatterplot(
    data=df,
    x='Total_Cost_INR',
    y='Profit_INR',
    hue='ROI_pct',
    palette='coolwarm',
    alpha=0.8,
    s=60
)

# Add quadrant reference lines at zero profit and median cost
median_cost = df['Total_Cost_INR'].median()
plt.axhline(0, color='black', linestyle='--', linewidth=1, label='Break-even Line (Zero Profit)')
plt.axvline(median_cost, color='gray', linestyle=':', linewidth=1, label='Median Cost Threshold')

# Add titles and labels
plt.title('Farm Risk & ROI Quadrant Analysis: Total Cost vs. Profit', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Total Cost (INR)', fontsize=12)
plt.ylabel('Profit (INR)', fontsize=12)

# Customize legend and layout
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='ROI (%)')
plt.tight_layout()

# Save and display the plot
plt.savefig('farm_roi_matrix.png', dpi=300)
plt.show()
```

**Outlier Investigation using Interquartile Range (IQR)**

```python
numeric_cols = ['Yield_Tonnes_Ha', 'Total_Cost_INR', 'Profit_INR', 'Rainfall_mm']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: Found {len(outliers)} potential outliers (Out of {len(df)} rows)")

# Visualize outliers with boxplots
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(y=df['Yield_Tonnes_Ha'], color='skyblue')
plt.title('Yield Outlier Distribution')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['Profit_INR'], color='salmon')
plt.title('Profit Outlier Distribution')
plt.tight_layout()
plt.show()
```

```python
# Define key numerical columns to analyze for outliers
num_cols = ['Yield_Tonnes_Ha', 'Total_Cost_INR', 'Profit_INR', 'Rainfall_mm', 'Fertilizer_kg_ha', 'Water_Used_m3']
outlier_summary = []

for col in num_cols:
    series = df[col].dropna()
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers beyond the 1.5 * IQR bounds
    outliers = series[(series < lower_bound) | (series > upper_bound)]

    outlier_summary.append({
        'Metric / Column': col,
        'Q1 (25%)': round(Q1, 2),
        'Q3 (75%)': round(Q3, 2),
        'IQR': round(IQR, 2),
        'Lower Bound': round(lower_bound, 2),
        'Upper Bound': round(upper_bound, 2),
        'Outlier Count': len(outliers),
        'Outlier %': round((len(outliers) / len(series)) * 100, 2)
    })

# Convert to DataFrame for neat display
outlier_df = pd.DataFrame(outlier_summary)
display(outlier_df)
```

## Comprehensive Project Conclusion & Agricultural Insights

The comprehensive exploratory data analysis, statistical evaluation, and multi-variable modeling of the agricultural performance dataset successfully deliver granular, data-driven insights into farm-level productivity, cost structures, and profitability drivers. By evaluating 4,000 agricultural records across multiple states, districts, seasons, and irrigation topologies, the project establishes a robust framework for understanding modern agricultural performance dynamics.

### Summary of Key Findings

* **Production Volume Polarization:** Production volume is heavily dominated by commercial and high-density biomass crops like Sugarcane, which accounts for the vast majority of total national production tonnage due to exceptionally high yields per hectare compared to staple cereals and pulses.


* **Financial Profitability Disparity:** A striking divergence exists between production volume and net profitability. While staple crops like Wheat, Rice, and Maize often operate near break-even or record negative net margins due to high operational expenditures, specialized commercial crops like Sugarcane and Chilli achieve high median profitability and strong financial returns.


* **Operational Cost Burdens:** Labor, land preparation, and machinery overhead constitute the primary components of total farm expenditure, significantly outweighing variable chemical costs such as fertilizers and pesticides. Managing baseline operational costs is critical for preventing financial deficits.


* **Irrigation and Resource Efficiency:** Irrigation methodologies profoundly influence risk and financial return distributions. Advanced techniques like Drip and Sprinkler systems demonstrate superior water-use efficiency (`Water_Efficiency_t_per_1000m3`) compared to traditional flood irrigation, mitigating vulnerability to seasonal rainfall variations.


* **Seasonal Performance Patterns:** Seasonal groupings (`Kharif`, `Rabi`, and `Zaid`) highlight distinct variations in average yield and net profit, with Kharif cycles generally benefiting from favorable monsoon rainfall distributions and higher aggregate returns.



### Strategic & Evidence-Based Recommendations

* **Adoption of Precision Irrigation:** Farming operations and agricultural planners should accelerate transitions toward micro-irrigation systems (Drip and Sprinkler) to optimize water-use efficiency and safeguard profit margins against unpredictable water shortages.


* **Cost-Structure Optimization:** Producers operating in low-margin staple crop categories must implement cost-control measures and leverage localized cooperative pooling for machinery and labor to improve net return on investment (ROI).


* **Data-Driven Crop Planning:** Agricultural policy frameworks should support regional diversification toward high-ROI, water-efficient crops while deploying safety nets or subsidies for essential staple food production in vulnerable agro-climatic zones.



### Project Limitations

* **Proxy and Synthetic Estimation Dependencies:** Certain economic variables—such as itemized labor versus equipment depreciation costs—rely on simulated financial distributions rather than real-time transactional accounting.


* **Geographical Generalization Limits:** The scope is constrained to specific districts and states within the dataset, meaning findings require localized calibration before scaling to entirely different agro-ecological zones.


* **Exclusion of External Market Shocks:** Macroeconomic price volatility, sudden export/import policy shifts, and extreme localized pest epidemics are modeled via risk percentages rather than dynamic time-series shocks.
